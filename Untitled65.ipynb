```python
!pip install gensim

import gensim
from gensim.models import Word2Vec
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('df_total.csv')
df.head()
```

```python
corpus_simple = []

frases_estudiante_pos = [
    "el estudiante es responsable y trabaja bien",
    "la estudiante es dedicada y muy responsable",
    "el estudiante hizo un buen trabajo en el proyecto",
    "la estudiante entregó todo a tiempo y con buena calidad",
    "el estudiante mostró esfuerzo y interés real",
    "la estudiante participa en clase y ayuda al grupo",
    "el grupo valora al estudiante responsable",
    "la profesora dijo que la estudiante es muy aplicada",
    "el profesor comentó que el estudiante es muy dedicado",
    "la estudiante tiene actitud positiva y respeta al equipo"
]

frases_estudiante_neg = [
    "el estudiante es vago y no participa",
    "la estudiante no entrega nada a tiempo",
    "el estudiante no muestra interés en clase",
    "la estudiante hizo un trabajo flojo y apurado",
    "el equipo piensa que el estudiante es irresponsable",
    "la estudiante se distrae y no colabora",
    "el estudiante llega tarde y no avisa",
    "la estudiante evita las tareas difíciles",
    "el estudiante no prepara las actividades",
    "la estudiante ignora las indicaciones del profesor"
]

frases_docente = [
    "el profesor dio una devolución clara al estudiante",
    "la profesora intentó apoyar al grupo",
    "el profesor habló con el estudiante sobre su rendimiento",
    "la profesora reconoció el buen esfuerzo del grupo",
    "el profesor pidió más compromiso al estudiante",
    "la profesora dijo que el trabajo fue correcto",
    "el profesor señaló que el trabajo fue muy bueno",
    "la profesora marcó problemas de actitud en clase",
    "el profesor resaltó el respeto y la colaboración",
    "la profesora destacó la calidad del proyecto final"
]

for _ in range(20):
    corpus_simple.extend(frases_estudiante_pos)
    corpus_simple.extend(frases_estudiante_neg)
    corpus_simple.extend(frases_docente)

```

```python
sentences = [frase.lower().split() for frase in corpus_simple]
print(sentences[0])
```

```python
model = Word2Vec(
    sentences,
    vector_size=50,  # tamaño de los vectores
    window=3,        # contexto de palabras vecinas
    min_count=1,     # incluye todas las palabras
    workers=4,       # cantidad de núcleos de CPU
    sg=1             # usa el modelo Skip-Gram
)

print(model.wv["estudiante"])
print(len(model.wv["estudiante"]))  # debería ser 50
```

```python
palabras_a_graficar = [
    "estudiante", "responsable", "vago", "esfuerzo",
    "calidad", "irresponsable", "colabora", "compromiso",
    "respeto", "actitud"
]

palabras_validas = [p for p in palabras_a_graficar if p in model.wv]
print("Palabras que se van a graficar:", palabras_validas)
```

```python
vectores = [model.wv[p] for p in palabras_validas]

pca = PCA(n_components=2)
coords_2d = pca.fit_transform(vectores)

plt.figure(figsize=(8,6))
x_vals = coords_2d[:, 0]
y_vals = coords_2d[:, 1]

plt.scatter(x_vals, y_vals)

for i, palabra in enumerate(palabras_validas):
    plt.annotate(palabra, (x_vals[i], y_vals[i]))

plt.title("Mapa 2D de embeddings (PCA) - Relaciones entre términos")
plt.xlabel("Componente 1")
plt.ylabel("Componente 2")
plt.grid(True)
plt.show()
```
